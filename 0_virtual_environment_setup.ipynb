# Virtual Environment Setup for AI Projects

This notebook guides you through the process of setting up a Python virtual environment named `ai_env`, activating it, and installing dependencies from a `requirements.txt` file. This is a crucial step for managing project-specific dependencies and avoiding conflicts. NOTE: All of the following commands have to be executed in the terminal of vscode

## 1. Create the Virtual Environment

We will use the `venv` module, which is part of the Python standard library, to create a new virtual environment named `ai_env`. This command will create a new directory `ai_env` in your current working directory, containing a Python interpreter and necessary files to isolate your project dependencies.

```python
python -m venv ai_env
```

## 2. Prepare `requirements.txt`

A `requirements.txt` file lists all the Python packages your project depends on. In a real project, you would generate this file using `pip freeze > requirements.txt` after installing your dependencies.

## 3. Install Dependencies 

pip install -r requirements.txt

## 4. Verify Installed Packages

Finally, let's verify that the packages listed in `requirements.txt` have been successfully installed within our `ai_env` virtual environment. This command will list all packages installed in the currently active kernel's environment.

```python
pip list
```
